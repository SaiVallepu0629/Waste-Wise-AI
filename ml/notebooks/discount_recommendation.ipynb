```python

import pandas as pd
import numpy as np
from pathlib import Path

# Your project folder
project = Path(
    r"C:\Users\srinu\Desktop\sai vallepu"
)

# Find the CSV file
files = list(
    project.rglob("discount_scenarios_simulated.csv")
)

if not files:
    raise FileNotFoundError(
        "discount_scenarios_simulated.csv not found"
    )

discounts = pd.read_csv(files[0])

print("Discount dataset loaded successfully!")
print("Shape:", discounts.shape)
print("\nColumns:")
print(discounts.columns.tolist())

display(discounts.head())
```

```python


print("===== DISCOUNT DATASET =====")

for col in discounts.columns:
    print(col, ":", discounts[col].dtype)

print("\nPossible discount columns:")
print([
    col for col in discounts.columns
    if any(word in col.lower()
           for word in ["discount", "price"])
])

print("\nPossible sales columns:")
print([
    col for col in discounts.columns
    if any(word in col.lower()
           for word in ["sold", "revenue", "demand"])
])

display(discounts.head(10))
```

```python

print("Total rows:", len(discounts))

print("\nMissing values:")
print(discounts.isnull().sum())

print("\nDuplicate rows:")
print(discounts.duplicated().sum())

print("\nNumeric summary:")
display(discounts.describe())
```
