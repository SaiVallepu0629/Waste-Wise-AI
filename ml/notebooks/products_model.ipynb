```python

import pandas as pd
from pathlib import Path

data_folder = Path(
project_root = Path.cwd().resolve()
while not (project_root / "ml" / "data" / "raw").is_dir():
    if project_root.parent == project_root:
        raise FileNotFoundError("Could not find the WasteWise-AI project root.")
    project_root = project_root.parent
data_folder = project_root / "ml" / "data" / "raw"
)

products = pd.read_csv(
    data_folder / "products.csv"
)

print("Products Dataset Loaded Successfully!")
print("Rows:", products.shape[0])
print("Columns:", products.shape[1])

display(products.head())
```

```python

print("Column Names:")
print(products.columns.tolist())

print("\nDataset Information:")
products.info()

print("\nStatistical Summary:")
display(products.describe(include="all"))
```

```python

print("Missing Values:")
print(products.isnull().sum())

print("\nMissing Value Percentage:")
print((products.isnull().mean() * 100).round(2))
```

```python

print("Duplicate Rows:", products.duplicated().sum())

# Create a cleaned copy
products_cleaned = products.drop_duplicates().copy()

print("Rows After Cleaning:", len(products_cleaned))
```

```python

inventory = pd.read_csv(
    data_folder / "perishable_inventory (1).csv"
)

batches = pd.read_csv(
    data_folder / "inventory_batches.csv"
)

print("Products and Inventory:")
print(sorted(set(products.columns) & set(inventory.columns)))

print("\nProducts and Batches:")
print(sorted(set(products.columns) & set(batches.columns)))
```

```python

output_folder = project_root / "ml" / "data" / "processed"

output_folder.mkdir(parents=True, exist_ok=True)

products_cleaned.to_csv(
    output_folder / "products_cleaned.csv",
    index=False
)

print("Cleaned Products Dataset Saved Successfully!")
```
