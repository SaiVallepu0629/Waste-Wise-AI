```python

import pandas as pd
from pathlib import Path

data_folder = Path(
project_root = Path.cwd().resolve()
while not (project_root / "ml" / "data" / "raw").is_dir():
    if project_root.parent == project_root:
        raise FileNotFoundError("Could not find the WasteWise-AI project root.")
    project_root = project_root.parent
data_folder = project_root / "ml" / "data" / "raw"
)

# Load inventory batches
batches = pd.read_csv(
    data_folder / "inventory_batches.csv"
)

print("Dataset Loaded Successfully!")
print("Rows:", batches.shape[0])
print("Columns:", batches.shape[1])

display(batches.head())
```

```python

# Display column names
print("Column Names:")
print(batches.columns.tolist())

# Dataset information
print("\nDataset Information:")
batches.info()

# Missing values
print("\nMissing Values:")
print(batches.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:")
print(batches.duplicated().sum())

# Statistical summary
display(batches.describe(include="all"))
```

```python

# Find expiry and date columns
date_columns = [
    col for col in batches.columns
    if "date" in col.lower()
    or "expiry" in col.lower()
    or "expire" in col.lower()
]

print("Date-related columns:", date_columns)

for col in date_columns:
    print(f"\n{col}")
    print(batches[col].head())
```

```python

inventory = pd.read_csv(
    data_folder / "perishable_inventory (1).csv"
)

common_columns = set(inventory.columns) & set(batches.columns)

print("Common Columns:")
print(sorted(common_columns))
```
